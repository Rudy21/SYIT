**Parameter Passing Mechanisms**

Python parameters are highly flexible, adapting to both positional requirements and dynamic key-value inputs.

**Positional vs. Keyword Arguments**

* **Positional Arguments:** Bound to the parameters in the exact order specified during definition.

* **Keyword Arguments:** Explicitly map arguments to parameters using the name=value format during the function call, allowing the order to be ignored.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

# Positional call (Order matters)
describe_pet("Dog", "Buddy")   # Output: I have a Dog named Buddy.

# Keyword call (Order does not matter)
describe_pet(pet_name="Whiskers", animal_type="Cat") # Output: I have a Cat named Whiskers.
```

**Default Parameters**

Default parameters allow parameters to assume a fallback value if omitted during the function call.

**Critical Rule:** Default arguments must follow non-default arguments in the function definition signature to prevent parsing ambiguity.

```python
# Correct Signature
def greet_user(username, greeting="Hello"):
    return f"{greeting}, {username}!"

print(greet_user("Amit"))             # Output: Hello, Amit!
print(greet_user("Amit", "Welcome"))  # Output: Welcome, Amit!
```

**Variable-Length Arguments (*args and kwargs)***

* ***args (Non-keyword variable arguments):** Packs an arbitrary number of trailing positional arguments into a tuple.

* **kwargs (Keyword variable arguments):** Packs an arbitrary number of named/keyword arguments into a dictionary.

```python
def analyze_classroom(instructor, *students, **metadata):
    print(f"Instructor: {instructor}")
    print(f"Students Tuple: {students}")
    print(f"Metadata Dict: {metadata}")

# Executing with dynamic inputs
analyze_classroom("Dr. Roy", "Alice", "Bob", "Charlie", Room=402, Subject="Python")
# Output:
# Instructor: Dr. Roy
# Students Tuple: ('Alice', 'Bob', 'Charlie')
# Metadata Dict: {'Room': 402, 'Subject': 'Python'}
```

**Return Values**

If no explicit return statement is written, or if a bare return is reached, Python implicitly returns None. Python can also return multiple values natively by separating them with commas, which automatically packs them into a single tuple.



```python
def get_min_max(numbers_list):
    minimum = min(numbers_list)
    maximum = max(numbers_list)
    return minimum, maximum # Returns a tuple

low, high = get_min_max([23, 5, 89, 43, 12])
print(f"Min: {low}, Max: {high}") # Output: Min: 5, Max: 89
```

**Scope Resolution (LEGB Rule)**

Python resolves variable references using the strict hierarchical LEGB order:

* **L (Local):** Names assigned within a function and not declared global.

* **E (Enclosing):** Names in the local scope of any and all enclosing functions (nested structures).

* **G (Global):** Names assigned at the top-level of a module file or declared global inside a function.

* **B (Built-in):** Predefined names in Python (e.g., open, range, print).

```python
x = "Global X"

def outer_function():
    x = "Enclosing X"

    def inner_function():
        global x
        # Modifying the global x, completely bypassing local/enclosing
        x = "Modified Global X"

    inner_function()
    print("Outer:", x)

outer_function()
print("Global:", x)
```

###**List Comprehension**

In Python, we frequently transform, clean, or filter sequences. The traditional approach relies on initializing an empty list and using a for loop to append elements.

While functional, this approach requires multiple lines of code, manages an external state (the empty list), and is slower due to the overhead of the .append() method. List comprehensions solve this by offering a declarative, optimized, single-line syntax.

**The Anatomy of a List Comprehension**
Every list comprehension maps directly to a traditional loop.

image.png

**Basic Transformations (Mapping)**
Mapping means taking an input sequence and applying a mathematical or structural change to every single element.

```python
#Traditional for loop:

celsius_temperatures = [0, 10, 20, 30, 40]
fahrenheit_temperatures = []

for temp in celsius_temperatures:
  if temp > 25:
    fahrenheit_temperatures.append((temp * 9/5) + 32)

print(fahrenheit_temperatures)
```

```python
fahrenheit_temperatures = [(temp * 9/5) + 32 for temp in celsius_temperatures if temp > 25]
print(fahrenheit_temperatures)
```

```python
fahrenheit_temperatures = [(temp * 9/5) + 32 for temp in celsius_temperatures]
print(fahrenheit_temperatures)
```

```python
fahrenheit_temperatures = [(temp * 9/5) + 32 for temp in celsius_temperatures]
```

**Conditional Filtering (if Statement)**

You can append an if statement to the very end of the comprehension to drop unwanted elements. If the condition evaluates to False, the item is skipped entirely.

```python
dataset = ["apple", "banana", "cherry", "date", "fig", "grape"]

# Keep only fruits that contain the letter 'a'
fruits_with_a = [fruit for fruit in dataset if 'a' in fruit]
```

**Transformation with if-else (Ternary Expression)**

When you need to transform elements based on a condition (rather than filtering them out), place a ternary if-else expression before the for clause:

```python
scores = [45, 88, 72, 30, 95]
status = []

for score in scores:
  if score >=50:
    status.append("Pass")
  else:
    status.append("Fail")

print(status)
```

```python
scores = [45, 88, 72, 30, 95]

# Map pass/fail labels
status = ["Pass" if score >= 50 else "Fail" for score in scores]
# Output: ['Fail', 'Pass', 'Pass', 'Fail', 'Pass']
```

**Nested Loops & Flattening Multi-Dimensional Data**

You can nest multiple for clauses. The order of for clauses in a list comprehension matches the order of standard nested for loops.

###**Generator Expressions ("Generator Comprehensions")**
A Generator Expression is a compact and memory-efficient way to create a generator object in Python. It is similar to a list comprehension, but instead of creating and storing all values in memory at once, it generates the values one at a time whenever they are needed.

This process is called **lazy evaluation.**

A generator expression is useful when working with large datasets because it saves memory and improves performance.

**Internal Mechanism & Memory Savings**
* **Eager Evaluation (List Comprehension):** Allocates memory for all $N$ items upfront.
* **Lazy Evaluation (Generator Expression):** Allocates a single lightweight frame object holding execution state; computes values only when requested.

**Why Generator Expressions?**

```python
#Suppose we want to generate numbers from 1 to 1 crore.


numbers = [x for x in range(10000000)]
'''
Here,

1. Python creates all 1 crore numbers immediately.
2. Large amount of memory is occupied.
3. Program becomes slower.'''
```

```python
#Using Generator Expression

numbers = (x for x in range(10000000))

'''
Here,

1. No numbers are created initially.
2. Values are generated only when requested.
3. Memory consumption is very small.
'''
```

**Syntax**

(**expression** for **item** in **iterable**)

```python
#For e.g.

numbers = (x for x in range(5))

print(numbers)
```

###**How Regex Works**

**Suppose we have an input:**

My roll number is 101.

**Regex Pattern**

\d+

**Meaning**

Find one or more digits.

**Output**

101

Regex searches the entire string and returns the matching pattern.

**Important Regex Functions**

Python provides several functions in the re module.

| Function        | Purpose                                                             |
| --------------- | ------------------------------------------------------------------- |
| `re.match()`    | Matches the pattern only at the beginning of the string             |
| `re.search()`   | Searches the first occurrence of the pattern anywhere in the string |
| `re.findall()`  | Returns all matching occurrences as a list                          |
| `re.finditer()` | Returns an iterator of all matches with position information        |
| `re.sub()`      | Replaces matching text                                              |
| `re.split()`    | Splits a string based on a pattern                                  |
| `re.compile()`  | Compiles a regex pattern for repeated use                           |


```python
#re.match()

#Matches only from the beginning of the string.


import re

text = "Python Programming"

result = re.match("Python", text)

print(result)

#To print the actual result:
print(result.group())

#If the pattern is

result_2 = re.match("Programming", text)

print(result_2.group())

#The output will be none because "Programming" is not at the beginning.
```

```python
#re.search()

#Searches the entire string.

import re

text = "I love Python Programming"

result = re.search("Python", text)

print(result.group())

#Unlike match(), search() checks the entire string.
```

```python
#re.findall()

#Returns every match.

import re

text = "Apple Mango Apple Banana Apple"

result = re.findall("Apple", text)

print(result)
```

```python
#re.finditer()

#Returns each match one by one along with its position.

import re

text = "Apple Mango Apple Banana"

result = re.finditer("Apple", text)

for item in result:
    print(item.start(), item.group())

#Useful when you need the location of each match.
```

```python
#re.sub()

#Used to replace matching text.

import re

text = "I like Java"

new_text = re.sub("Java", "Python", text)

print(new_text)
```

```python
#re.split()

#Splits text using a pattern.

import re

text = "Apple,Mango,Grapes"

result = re.split(",", text)

print(result)
```

```python
#Hierarchical Inheritance

class Vehicle:

  def start(self):
    print("Vehicle Started")

class Car(Vehicle):

  def drive(self):
    print("Car is Driving")

class Bike(Vehicle):
  def ride(self):
    print("Bike is Riding")

car = Car()
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
```

```python
#Real world Example: Banking System

class BankAccount:
  def __init__(self, account_number, name, balance):
    self.account_number = account_number
    self.name = name
    self.balance = balance

  def display_account(self):
    print("Account Number", self.account_number)
    print("Account Holder", self.name)
    print("Balance Amt", self.balance)

  def deposit(self, amount):
    self.balance += amount
    print("Rupees", amount, "Deposited Successfully")
    print("Current Balance: ", self.balance)

  def withdraw(self,amount):
    if amount <= self.balance:
      self.balance -= amount
      print("Rupees", amount, "Withdrawn Successfully")
      print("Current Balance: ", self.balance)
    else:
      print("Insufficient Balance")

#Child Class 1
#Single Inheritance

class SavingsAccount(BankAccount):

  def __init__(self,account_number, name, balance, interest_rate):
    super().__init__(account_number, name, balance)
    self.interest_rate = interest_rate

  def calculate_interest(self):
    interest = self.balance * self.interest_rate / 100
    print("Interest Earned: ", interest)

  #method overriding
  def display_account(self):
    print("\nSAVINGS ACCOUNT")
    print("Account Number: ", self.account_number)
    print("Account Holder: ", self.name)
    print("Balance: ",self.balance)
    print("Interest Rate: ", self.interest_rate, "%")

#Child Class 2
#Hierarchical inheritance

class CurrentAccount(BankAccount):
  def __init__(self,account_number,name,balance,overdraft_limit):
    super().__init__(account_number,name,balance)
    self.overdraft_limit = overdraft_limit

  def withdraw(self,amount):
    if amount <= self.balance + self.overdraft_limit:
      self.balance -= amount
      print("Rupees", amount, "withdrawn successfully")
      print("Current Balance: ", self.balance)
    else:
      print("Withdrawal exceeds overdraft limit")

  def display_account(self):
    print("CURRENT ACCOUNT")
    print("Account Number: ", self.account_number)
    print("Account Holder: ", self.name)
    print("Account Balance: ", self.balance)
    print("Overdraft Limit: ", self.overdraft_limit)

#Multilevel Inheritance

class PremiumSavingsAccount(SavingsAccount):
  def __init__(self, account_number,name,balance,interest_rate,reward_points):
    super().__init__(account_number,name,balance,interest_rate)
    self.reward_points = reward_points

  def display_rewards(self):
    print("Reward Points: ", self.reward_points)

  def add_reward_points(self,points):
    self.reward_points += points
    print("Reward Points Added: ",points)
    print("Total Reward Points: ", self.reward_points)


#Object Creation

print("BANKING SYSTEM")

#Object of BankAccount
account = BankAccount(
    "BA1001",
    "Rahul",
    100000
)

account.display_account()
account.deposit(1000)
account.withdraw(900)

#Object of SavingsAccount
savings = SavingsAccount(
    "SA1001",
    "HowToTyping",
    20000,
    6
)

savings.display_account()
savings.deposit(5000)
savings.calculate_interest()

#Object of CurrentAccount
current = CurrentAccount(
    "CA1001",
    "AppDeveloper",
    30000
)

current.display_account()
current.withdraw(20000)

#Object of PremiumSavingsAccount
premium = PremiumSavingsAccount(
    "PSA1001",
    "Developer",
    40000
)

premium.display_account()
premium.calculate_interest()
premium.add_reward_points(50)
premium.display_rewards()
```
